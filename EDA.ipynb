```python
import pandas as pd
import numpy as np

# Sample DataFrame with missing values
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda', 'James'],
    'Age': [28, np.NAN, 35, 32, np.NAN],
    'Salary': [72000, 68000, np.nan, 79000, 85000],
    'Department': ['IT', 'HR', np.NAN, 'Finance', 'IT']
}

df = pd.DataFrame(data)

# 1. Display the DataFrame with missing values
print("Original DataFrame:")
print(df)

```

```python
# 2. Detect missing values
print("\nDetecting missing values:")
print(df.isnull())

```

```python
# 3. Count missing values in each column
print("\nCounting missing values in each column:")
print(df.isnull().sum())

```

```python
# 4. Drop rows with any missing values
df_dropna = df.dropna()
print("\nDataFrame after dropping rows with any missing values:")
print(df_dropna)

```

```python
# 5. Drop columns with any missing values
df_dropna_columns = df.dropna(axis=1)
print("\nDataFrame after dropping columns with any missing values:")
print(df_dropna_columns)

```

```python
# 6. Fill missing values with a specified value
df_fillna = df.fillna(value={'Age': df['Age'].mean(), 'Salary': df['Salary'].median(), 'Department': 'Unknown'})
print("\nDataFrame after filling missing values with specific values:")
print(df_fillna)

```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda', 'James'],
    'Age': [28, np.NAN, 35, 32, np.NAN],
    'Salary': [72000, 68000, np.nan, 79000, 85000],
    'Department': ['IT', 'HR', np.NAN, 'Finance', 'IT']
}

df = pd.DataFrame(data)

# 1. Display the DataFrame with missing values
print("Original DataFrame:")
print(df)


# 7. Forward fill (propagate last valid observation forward to next)
df_ffill = df.fillna(method='ffill')
print("\nDataFrame after forward filling missing values:")
print(df_ffill)

```

```python
# 8. Backward fill (propagate next valid observation backward to previous)
df_bfill = df.fillna(method='bfill')
print("\nDataFrame after backward filling missing values:")
print(df_bfill)

```

```python
# 9. Interpolation to estimate missing values
df_interpolate = df.interpolate()
print("\nDataFrame after interpolating missing values:")
print(df_interpolate)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Creating a sample dataset
data = {
    'Age': [23, 25, 22, 30, 28, 35, 40, 23, 24, 22],
    'Height': [165, 170, 160, 175, 180, 172, 168, 158, 165, 170],
    'Weight': [65, 70, 60, 75, 80, 72, 68, 58, 65, 70],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male'],
    'Smoker': ['No', 'Yes', 'No', 'No', 'Yes', 'No', 'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

# Step 1: Understand the Data
print("Data Types:\n", df.dtypes)
print("\nSummary Statistics:\n", df.describe())

# Step 2: Univariate Analysis - Numerical Data
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.histplot(df['Age'], kde=True, color='blue')
plt.title('Age Distribution (Discrete Data)')
     
plt.subplot(1, 2, 2)
sns.histplot(df['Height'], kde=True, color='green')
plt.title('Height Distribution (Continuous Data)')

plt.tight_layout()
plt.show()

# Step 3: Univariate Analysis - Categorical Data
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.countplot(x='Gender', data=df, palette='Set1')
plt.title('Gender Count (Categorical Data)')

plt.subplot(1, 2, 2)
sns.countplot(x='Smoker', data=df, palette='Set2')
plt.title('Smoker Count (Categorical Data)')

plt.tight_layout()
plt.show()

# Step 4: Bivariate Analysis - Numerical Data
sns.scatterplot(x='Age', y='Weight', hue='Gender', data=df)
plt.title('Scatter Plot of Age vs Weight')
plt.show()

# Step 5: Outlier Detection - Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(x='Gender', y='Height', data=df)
plt.title('Boxplot of Height by Gender')
plt.show()

# Step 6: Feature Engineering
# Let's create a new feature: BMI (Body Mass Index)
df['BMI'] = df['Weight'] / (df['Height'] / 100) ** 2
print("\nData with new BMI feature:\n", df)

# Step 7: Visualization of the new feature
sns.histplot(df['BMI'], kde=True, color='purple')
plt.title('BMI Distribution')
plt.show()
```
